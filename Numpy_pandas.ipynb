```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Multiplication:", a * b)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard Deviation:", np.std(a))
print("Sum:", np.sum(a))
```

```python
import numpy as np

a = np.array([12, 45, 23, 67, 10])

print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
```

```python
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

print(a)
print("Shape:", a.shape)
print("Dimensions:", a.ndim)
```

```python
import numpy as np

a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print("Addition:")
print(a + b)

print("Multiplication:")
print(np.dot(a, b))
```

```python
import pandas as pd

a = pd.Series([10, 20, 30, 40])

print(a)
```

```python
import pandas as pd

data = {
    'Name': ['Rahul', 'Anu', 'Ravi'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Marks': [80, 90, 70, 85, 95]
})

print(df.head(3))
print(df.tail(2))
```

```python
import pandas as pd

df = pd.DataFrame({
    'Marks': [80, 90, 70, 80, 100]
})

print("Mean:", df['Marks'].mean())
print("Median:", df['Marks'].median())
print("Mode:", df['Marks'].mode())
print("Sum:", df['Marks'].sum())
```

```python
import pandas as pd

df = pd.DataFrame({
    'Age': [18, 19, 20, 21, 22],
    'Marks': [70, 80, 90, 85, 95]
})

print(df.describe())
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Marks': [80, np.nan, 90]
})

print(df.isnull())

df['Marks'] = df['Marks'].fillna(0)

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Maths': [80, 90, 70, 85],
    'Science': [75, 95, 65, 80]
})

print("Correlation:")
print(df.corr())

print("Covariance:")
print(df.cov())
```
